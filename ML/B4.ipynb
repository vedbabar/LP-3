```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x + 3)**2
def derivative(x):
    return 2 * (x + 3)
```

```python
x = 2
learning_rate = 0.1
iterations = 500
```

```python
for i in range(iterations):
    grad = derivative(x)
    x = x - learning_rate * grad

    print("Iteration:", i+1, "x:", x, "y:", function(x))
```

```python
print("Minimum value of x:", x)

print("Minimum value of y:", function(x))
```

```python
x_values = np.linspace(-6, 2, 100)

y_values = function(x_values)

plt.plot(x_values, y_values)

plt.xlabel("x")
plt.ylabel("y")

plt.title("Graph of y = (x + 3)^2")

plt.grid()

plt.show()
```
